```python
import os
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq

#  Load updated LLM (NEW MODEL)
llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.5
)

#  Take notes input
def load_notes():
    print("Paste your notes (type END to finish):\n")
    lines = []
    while True:
        line = input()
        if line.strip() == "END":
            break
        lines.append(line)
    return "\n".join(lines)


#  Generate summary
def generate_summary(text):
    prompt = f"""
Summarize the following notes in simple and short points:

{text}
"""
    return llm.invoke(prompt).content


#  Generate Q&A
def generate_questions(text):
    prompt = f"""
Generate 5 important questions with answers from the following notes:

{text}
"""
    return llm.invoke(prompt).content


# Generate MCQ Quiz
def generate_quiz(text):
    prompt = f"""
Create 5 MCQ questions with 4 options and highlight the correct answer:

{text}
"""
    return llm.invoke(prompt).content


#  Main program
def main():

    notes = load_notes()

    if not notes.strip():
        print(" No input provided!")
        return

    print("\n Processing...\n")

    try:
        summary = generate_summary(notes)
        questions = generate_questions(notes)
        quiz = generate_quiz(notes)

        print("\n SUMMARY:\n")
        print(summary)

        print("\n QUESTIONS & ANSWERS:\n")
        print(questions)

        print("\n QUIZ:\n")
        print(quiz)

    except Exception as e:
        print(" Error:", e)


if __name__ == "__main__":
    main()
```

```python
import os
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq

# Load LLM
llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.3
)

# Take code input
def get_code():
    print("\nPaste your code (type END to finish):\n")
    lines = []
    while True:
        line = input()
        if line.strip() == "END":
            break
        lines.append(line)
    return "\n".join(lines)


# Explain
def explain_code(code):
    prompt = f"Explain the following code line by line:\n{code}"
    return llm.invoke(prompt).content


# Find bugs
def find_bugs(code):
    prompt = f"Find errors in this code and explain them:\n{code}"
    return llm.invoke(prompt).content


# Fix code
def fix_code(code):
    prompt = f"Fix the errors and give corrected code:\n{code}"
    return llm.invoke(prompt).content


# Optimize
def optimize_code(code):
    prompt = f"Optimize this code for better performance:\n{code}"
    return llm.invoke(prompt).content


# Menu system
def main():

    code = get_code()

    if not code.strip():
        print("⚠️ No code provided!")
        return

    while True:
        print("\n===== SELECT OPTION =====")
        print("1. Explain Code")
        print("2. Find Bugs")
        print("3. Fix Code")
        print("4. Optimize Code")
        print("5. Exit")

        choice = input("Enter your choice: ")

        try:
            if choice == "1":
                print("\n📘 Explanation:\n", explain_code(code))

            elif choice == "2":
                print("\n🐞 Bugs:\n", find_bugs(code))

            elif choice == "3":
                print("\n✅ Fixed Code:\n", fix_code(code))

            elif choice == "4":
                print("\n⚡ Optimized Code:\n", optimize_code(code))

            elif choice == "5":
                print("Exiting...")
                break

            else:
                print("❌ Invalid choice!")

        except Exception as e:
            print("❌ Error:", e)


if __name__ == "__main__":
    main()
```
